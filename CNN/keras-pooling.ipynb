```python
from keras.layers import Dense, Conv2D,Flatten, MaxPooling2D
from keras import Sequential
from keras.datasets import mnist 
```

```python
(X_train,y_train),(X_test,y_test)=mnist.load_data()
```

```python
model = Sequential()  # Create sequential CNN model

model.add(Conv2D(32, kernel_size=(3,3),padding='valid', activation='relu',input_shape=(28,28,1)))
# 32 filters, 3×3 kernel, input = 28×28 grayscale image
model.add(MaxPooling2D(pool_size=(2,2), strides=2, padding='valid'))
# Takes maximum value from each 2×2 region
# Reduces image siz
model.add(Conv2D(32, kernel_size=(2,2),strides=2,padding='valid',activation='relu'))
# Second convolution layer
model.add(MaxPooling2D(pool_size=(2,2),strides=2, padding='valid'))
# Further reduces feature map size
model.add(Flatten())
# Converts 2D feature maps into 1D vector
model.add(Dense(128, activation='relu'))
# Fully connected layer with 128 neurons
model.add(Dense(10, activation='softmax'))
# Output layer: 10 classes (digits 0–9)
model.summary()
# Displays model architecture and number of parameters
```
