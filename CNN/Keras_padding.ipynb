```python
import tensorflow                 # Main deep learning library
from tensorflow import keras       # High-level TensorFlow API

from keras.layers import Dense, Conv2D, Flatten  # Neural network layers
from keras import Sequential                    # Build model layer-by-layer
from keras.datasets import mnist                # MNIST handwritten digits dataset
```

```python
(X_train,y_train), (X_test,y_test)=mnist.load_data()
```

```python
model = Sequential()  # Create a sequential CNN model
model.add(Conv2D(32, kernel_size=(3,3), padding='valid',
                 activation='relu', input_shape=(28,28,1)))
# 32 filters, 3×3 kernel (Size) each filter, ReLU activation padding valid means it wont use 
model.add(Conv2D(32, kernel_size=(3,3), padding='valid',
                 activation='relu'))
# Second convolution layer with 32 filters
model.add(Conv2D(32, kernel_size=(3,3), padding='valid',
                 activation='relu'))
# Third convolution layer with 32 filters
model.add(Flatten())
# Convert 2D feature maps into 1D vector
model.add(Dense(128, activation='relu'))
# Fully connected layer with 128 neurons
model.add(Dense(10, activation='softmax'))
# Output layer: 10 classes (digits 0–9)
model.summary()
```

```python
model = Sequential()  # Create a sequential CNN model
model.add(Conv2D(32, kernel_size=(3,3), padding='same',
                 activation='relu', input_shape=(28,28,1)))
# 32 filters, 3×3 kernel (Size) each filter, ReLU activation padding same means it  use 0 adding padding  
model.add(Conv2D(32, kernel_size=(3,3), padding='same',
                 activation='relu'))
# Second convolution layer with 32 filters
model.add(Conv2D(32, kernel_size=(3,3), padding='same',
                 activation='relu'))
# Third convolution layer with 32 filters
model.add(Flatten())
# Convert 2D feature maps into 1D vector
model.add(Dense(128, activation='relu'))
# Fully connected layer with 128 neurons
model.add(Dense(10, activation='softmax'))
# Output layer: 10 classes (digits 0–9)
model.summary()
```

```python
#Strides (horizontal vertical)movement Stride means how many pixels the filter/kernel moves at a time over an im
model = Sequential()  # Create a sequential CNN model
model.add(Conv2D(32, kernel_size=(3,3), padding='same',strides=(2,2),activation='relu', input_shape=(28,28,1)))
# 32 filters, 3×3 kernel (Size) each filter, ReLU activation padding same means it  use 0 adding padding  
model.add(Conv2D(32, kernel_size=(3,3), padding='same', strides=(2,2),activation='relu'))
# Second convolution layer with 32 filters
model.add(Conv2D(32, kernel_size=(3,3), padding='same',strides=(2,2),activation='relu'))
# Third convolution layer with 32 filters
model.add(Flatten())
# Convert 2D feature maps into 1D vector
model.add(Dense(128, activation='relu'))
# Fully connected layer with 128 neurons
model.add(Dense(10, activation='softmax'))
# Output layer: 10 classes (digits 0–9)
model.summary()
```
